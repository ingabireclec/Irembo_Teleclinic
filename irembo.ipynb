#### Null Handlings

Null values in the Lab Tests table were not dropped or filled because they carry clinical meaning. The 23 nulls in upload_datetime, tat_hours, and clinician_viewed correspond exactly to the 23 records where result_uploaded = No — structurally expected since a result that was never uploaded cannot have a turnaround time or clinician review. The 85 nulls in hours_to_view break into two groups: 23 from never-uploaded results and 62 from results the clinician never viewed. These 62 nulls are not missing data — they are the patient safety finding. Dropping them would have hidden the most clinically significant signal in the dataset. All TAT and review rate calculations use only uploaded results as the denominator.

```python
# Are there consultations with no matching records in clinical tables?
# A completed consultation should have SOMETHING — notes, lab, prescription, or referral

completed = consultation_df[consultation_df['status'] == 'Completed']

# Which completed consultations have no lab test?
no_lab = set(completed['consultation_id']) - set(lab_tests_df['consultation_id'])
print(f"Completed consultations with no lab test: {len(no_lab)}")

# Which completed consultations have no prescription?
no_rx = set(completed['consultation_id']) - set(prescriptions_df['consultation_id'])
print(f"Completed consultations with no prescription: {len(no_rx)}")

# Which completed consultations have no followup?
no_fu = set(completed['consultation_id']) - set(follow_ups_df['original_consult_id'])
print(f"Completed consultations with no followup: {len(no_fu)}")

# Which completed consultations have no referral?
no_ref = set(completed['consultation_id']) - set(referrals_df['consultation_id'])
print(f"Completed consultations with no referral: {len(no_ref)}")

# Most importantly — completed consultations with NOTHING at all
no_anything = (
    set(completed['consultation_id']) - 
    set(lab_tests_df['consultation_id']) - 
    set(prescriptions_df['consultation_id']) - 
    set(follow_ups_df['original_consult_id']) - 
    set(referrals_df['consultation_id'])
)
print(f"Completed consultations with no clinical output at all: {len(no_anything)}")
```
